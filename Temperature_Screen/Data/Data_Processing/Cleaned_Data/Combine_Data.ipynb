```python
import pandas as pd
from os import listdir
from os.path import isfile, join
import seaborn as sns
```

```python
def combine_data(file_path, absorbance):
    all_data = []
    counter = 0
    for f in listdir("./"):
        if f.endswith(".csv"):
            data = pd.read_csv(join("./", f))
            # data = data[data["Absorbance"] == absorbance]
            data.loc[data["Temperature"] == 25, "Temperature"] += counter
            counter += 0.1
            all_data.append(data)
                
    all_data = pd.concat(all_data)
    all_data = all_data[all_data.columns[1:]]
    return all_data.reset_index()
```

```python
all_data = combine_data("./", 680)
all_data.tail()
```

```python
#2343 rep 4 at 32°C wasn't filled, so we can remove it from the analysis

all_data.drop(list(all_data[(all_data.Cell_Line == "2342") & (all_data.Replicate == 4) & ((all_data.Temperature == 32) | (all_data.Temperature == 28) |(all_data.Temperature == 25.1))].index), inplace = True)
all_data[(all_data.Cell_Line == "2342") & (all_data.Replicate == 4) & (all_data.Temperature == 32)]
# all_data[(all_data.Cell_Line == "2342") & (all_data.Replicate == 4) & (all_data.Temperature == 20)]
```

```python
# Export Data
all_data["ID"] = all_data["Cell_Line"].astype("str") + "_"+ all_data["Temperature"].astype("str") +"_"+ all_data["Replicate"].astype("str") +"_"+ all_data["Absorbance"].astype("str")
data_export_reps = all_data.pivot_table(values='value', index='Hours', columns='ID')

data_export_reps.shape
data_export_reps.to_csv("../Annotating/Raw_Temperature_Screen.csv")
```

```python
import pandas as pd
from os import listdir
from os.path import isfile, join
```

```python
def get_abs(abs_df, plate_template, absorbance, line):
    '''
    Function to extract the particular absorbance of a certain cell line and return that as a list
    
    Variables:
        abs_df = the pandas dataframe that contains the absorbances of all cell lines
        plate_template = the positioning of the lines on each of the cell plates
        absorbance = well...pretty self explainatory 
        line = cell line that you're intersted in 
    '''
    
    #first reformat the last column of the spreadsheet to only contain the absorbance as a number
    abs_df["Unnamed: 14"] = pd.to_numeric(abs_df["Unnamed: 14"].replace("Read [123]:", "", regex = True))
    #Filter by absorbance
    filtered_df = abs_df.loc[abs_df["Unnamed: 14"] == absorbance].reset_index(drop=True)
    #Filter by line
    values = filtered_df.where(plate_template == line).apply(lambda x: x.dropna().to_list())
    #Filter to only get lists with items 
    if isinstance(values, pd.DataFrame):
        return [0,0,0,0,0,0,0,0]
    
    else:
        abs_list = values[values.str.len() > 0].to_list()
        #Flatten the list (i.e. condense it to 1 list)
        abs_list = sum(abs_list, [])
        return np.array(abs_list)
```

```python
cell_lines = [3059, 1691, 1690, 2342, 4532, 2344, 2343, 1952, 2931, "JG5", "BMS"] #2932
abs_lst = [680,750]
temperature = [25, 20, 30]

#Get the positions of the lines on their plates
plate_template = pd.read_excel("Plate_template.xlsx", sheet_name = 0)
```

```python
absorbance_dict = {}
temp_data = [53,104,155]# Line numbers of where the temperature data starts
#go through the excel files in the cwd
for f in listdir("./"):
    if f == "Plate_template.xlsx":
        continue
    print(f)
    f_name = f.strip().split(".")
    #only look at files that end in an xl format
    if isfile(join("./", f)) and (f_name[-1][:2] == "xl"):
        for index in [0,1,2]: 
            #go through every plate layout in the file (0 = 25°C, 1 = 30°C, 2 = 35°C)
            abs_df = pd.read_excel(f, usecols = "B:O", skiprows = temp_data[index], nrows = 48)
            for line in cell_lines:
                for absorb in abs_lst:
                    key = "{abso}_{temp}_{cell_line}".format(abso = absorb, temp = temperature[index], cell_line = line)
                    abs_data = get_abs(abs_df, plate_template, absorb, line)
                    if len(abs_data) >= 8:
                        absorbance_dict[f_name[0]+"_"+key] = list(abs_data)
                        if len(abs_data) != 8:
                            print(key)
```

```python
#Is the variation of cell lines at a given temperature and absorbance significant?
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Create the formula for the ANOVA test
formula = 'log10_transformed ~ C(Cell_Line) + C(Temperature) + C(Cell_Line):C(Temperature)'
# formula = 'log10_transformed ~ C(Cell_Line) + C(Temperature) + C(Absorbance) + C(Cell_Line):C(Temperature):C(Absorbance)'
# Use the above formula for df_filtered

#ANOVA model
model = ols(formula, data=df_filtered_680).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)

# More specifically, the term C(Cell_Line) being significant indicates that there are significant differences 
# in the log10_transformed values between different Cell_Line groups, even after controlling for the effect of 
# Temperature. The term C(Temperature) being significant indicates that there are significant differences in the 
# log10_transformed values between different Temperature levels, even after controlling for the effect of 
# Cell_Line. Finally, the term C(Cell_Line):C(Temperature) being significant indicates that the effect of 
# Cell_Line on log10_transformed varies depending on the level of Temperature, and vice versa, and that this 
# interaction is significant for absorbance at 680.

# Overall, these results suggest that both Cell_Line and Temperature are important factors in explaining the 
# variation in log10_transformed, and that their combined effect is not simply additive, but involves complex 
# interactions.
```

```python
from scipy.stats import kruskal

df_abs680 = data_sorted[data_sorted['Absorbance'] == 680]
variatn = []
variances_cl_temp = {}
for cell_line in df_abs680['Cell_Line'].unique():
    variances_temp = []
    for temp in df_abs680['Temperature'].unique():
        df_cl_temp = df_abs680[(df_abs680['Cell_Line'] == cell_line) & (df_abs680['Temperature'] == temp)]
        variances_temp.append([df_cl_temp['log10_transformed'].var()])
    variances_cl_temp[cell_line] = variances_temp
    variatn.append(sum(sum(variances_temp, []))/3)
    print(cell_line, sum(sum(variances_temp, []))/3)
    stat, p_value = kruskal(*variances_temp)
    print('Kruskal-Wallis test:')
    print(f'Statistic: {stat:.4f}')
    print(f'p-value: {p_value:.4f}')

stat, p_value = kruskal(*variances_cl_temp.values())

print('Kruskal-Wallis test:')
print(f'Statistic: {stat:.4f}')
print(f'p-value: {p_value:.4f}')
```
